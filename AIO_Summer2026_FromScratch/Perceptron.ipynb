```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

```python
X, y = make_blobs(
    n_samples=1000,
    n_features=2,
    centers=2,
    cluster_std=1.25,
    random_state=42
)
```

```python
def plot_perceptron(ax, X, y, W, b, title):
    # Data
    ax.scatter(
        X[:, 0],
        X[:, 1],
        c=y,
        alpha=0.6
    )

    # Decision boundary
    x1 = np.linspace(
        X[:, 0].min(),
        X[:, 0].max(),
        200
    )

    x2 = -(W[0] * x1 + b) / W[1]

    ax.plot(
        x1,
        x2,
        linewidth=3,
        label="Decision Boundary"
    )

    # W vector
    ax.quiver(
        0, 0,
        W[0], W[1],
        angles="xy",
        scale_units="xy",
        scale=1,
        width=0.01,
        label=f"W = {W}"
    )

    # Formatting
    ax.set_title(title)
    ax.set_xlabel("x1")
    ax.set_ylabel("x2")
    ax.axhline(0, linewidth=1)
    ax.axvline(0, linewidth=1)
    ax.set_aspect("equal")
    ax.legend()
```

```python
# Perceptron parameters
W = np.array([1,3])
W = W / np.linalg.norm(W)
b = 0
```

```python
X_real = X.copy()

X_norm = X / np.linalg.norm(
    X,
    axis=1,
    keepdims=True
)

fig, ax = plt.subplots(1, 2, figsize=(14, 6))

plot_perceptron(
    ax[0],
    X_real,
    y,
    W,
    b,
    "Real X"
)

plot_perceptron(
    ax[1],
    X_norm,
    y,
    W,
    b,
    "Normalized X"
)

plt.tight_layout()
plt.show()
```
